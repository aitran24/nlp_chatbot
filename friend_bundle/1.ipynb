```python
from qdrant_client import QdrantClient
from sentence_transformers import SentenceTransformer

client = QdrantClient(path="cache/qdrant_local_bench")  # Đảm bảo trùng với QDRANT_PATH trong config.py

# Test query thực tế
model2 = SentenceTransformer("BAAI/bge-m3", local_files_only=False)
q_vec = model2.encode("sĩ quan dự bị nhập ngũ bao nhiêu lâu", normalize_embeddings=True).tolist()

results = client.query_points(
    collection_name="chatbot_documents",
    query=q_vec,
    limit=3,
    with_payload=True,
).points

for r in results:
    print(f"Score: {r.score:.4f}")
    print(f"Text: {r.payload.get('text', '')[:200]}")
    meta = r.payload.get("metadata", {})
    print(f"Metadata: {meta}")
    print("---")
```
